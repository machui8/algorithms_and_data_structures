  # Question One: Check Duplicates in Tree

  Given the `root` of a binary tree, check whether it is contains a duplicate value. If a duplicate exists, return the duplicate value. If there are multiple duplicates, return the one with the closest distance to the root. If no duplicate exists, return -1.

#### Starter Code for Question 1

```python
from collections import deque

# Definition for a binary tree node
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate(root):
    if not root:
        return -1
    
    seen = set()
    
    # BFS (or Level Order Traversal)
    queue = deque([root])
    
    while queue:
        node = queue.popleft()
        
        # If the value seen before, it will be the closest duplicate
        if node.val in seen:
            return node.val
        
        # Add current value to the set of seen values
        seen.add(node.val)
        
        # Add child node to the queue to continue 
        # Repeat until the queue is empty
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
            
    return -1
```

```python
def build_tree(arr):
    if not arr: return None
    root = TreeNode(arr[0])
    queue = deque([root])
    i = 1
    while i < len(arr):
        curr = queue.popleft()
        if i < len(arr) and arr[i] is not None:
            curr.left = TreeNode(arr[i])
            queue.append(curr.left)
        i += 1
        if i < len(arr) and arr[i] is not None:
            curr.right = TreeNode(arr[i])
            queue.append(curr.right)
        i += 1
    return root

# Example 1
root_data_1 = [1, 2, 2, 3, 5, 6, 7]
root_node_1 = build_tree(root_data_1)
result_1 = is_duplicate(root_node_1)

print(f"Input: {root_data_1}")
print(f"Output: {result_1}")

# Example 2
root_data_2 = [1, 10, 2, 3, 10, 12, 12]
root_node_2 = build_tree(root_data_2)
result_2 = is_duplicate(root_node_2)

print(f"Input: {root_data_2}")
print(f"Output: {result_2}")

# Example 3
root_data_3 = [10, 9, 8, 7]
root_node_3 = build_tree(root_data_3)
result_3 = is_duplicate(root_node_3)

print(f"Input: {root_data_3}")
print(f"Output: {result_3}")
```

1 example from my partner
Move all zeros to the end of the list and keep the order for the remaining non-zero number
Input: nums = [0, 0, 0, 1]
Output: [1, 0, 0, 0]


-   Copy the solution your partner wrote. 


```python
# Solution from partner wrote
from typing import List
def move_zeros_to_end(nums: List[int]) -> List[int]:
    result = [num for num in nums if num != 0]  # keep non-zero elements
    zeros = [0] * (len(nums) - len(result))     # count and add zeros
    return result + zeros

print(move_zeros_to_end([0, 1, 0, 3, 12]))     # [1, 3, 12, 0, 0]
print(move_zeros_to_end([4, 0, 5, 0, 0, 6]))   # [4, 5, 6, 0, 0, 0]
```


-   Explain why their solution works in your own words.


The solution works. Firstly, it looks up all non-zero numbers from left to right and store into a list [result] in a preserved order.
Then it counts the number of zero by substracting the original list's length and the result list's. Then, it uses multiplication ([0] * count) to create a new list containing exactly the number of zeros. For example, if the original list had 5 items and the result has 3, it generates 2 zeros [0, 0]. Finally, it uses "+" to concatenate the two lists. The result list (non-zeros) goes first while the zeros list goes next. This ensures all non-zeros are at the front and all zeros are pushed to the end. 



-   Explain the problem’s time and space complexity in your own words.


Time Complexity: O(n)
It is because every element in the input must be "touched" at least once to determine if it is a zero or a non-zero number, the time complexity is O(n) 

Space Complexity: O(n)
It is because we create a new copy of the data rather than modify the original list in-place, the space complexity is O(n)


-   Critique your partner's solution, including explanation, and if there is anything that should be adjusted.


One of the critique is the space inefficiency. It is because the code creates 3 [result] [zeros] and the concatenated list. If a list with millions numbers, it will spend more memory or RAM to hold all. 

```python
# Suggestion
def move_zeros_to_end(nums: List[int]) -> None:
    # 'last_non_zero' tracks where the next non-zero should go
    last_non_zero = 0
    
    for current in range(len(nums)):
        if nums[current] != 0:
            # Swap elements
            nums[last_non_zero], nums[current] = nums[current], nums[last_non_zero]
            last_non_zero += 1

# Case 1: [0, 1, 0, 3, 12]
case_1 = [0, 1, 0, 3, 12]
move_zeros_to_end(case_1)
print(f"Result 1: {case_1}") 

# Case 2: [4, 0, 5, 0, 0, 6]
case_2 = [4, 0, 5, 0, 0, 6]
move_zeros_to_end(case_2)
print(f"Result 2: {case_2}") 
```
