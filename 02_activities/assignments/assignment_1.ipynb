## Question One: First Duplicate in List
**Description**  
Given a list of integers, return the **first value that appears more than once**. If there are multiple duplicates, return the one that appears **first** in the list. If no duplicate exists, return `-1`.

**Examples**
```python
Input: nums = [3, 1, 4, 2, 5, 1, 6]
Output: 1
```
```python
Input: nums = [7, 8, 9, 10]
Output: -1
```
```python
Input: nums = [4, 5, 6, 4, 6]
Output: 4
```

**Question 1 Starter Code**

````python
# Using a list

from typing import List

def list_first_duplicate(nums: List[int]) -> int:
    seen = []   
    for n in nums:
        if n in seen:   
            return n
        seen.append(n)
    return -1
````

````python
print(list_first_duplicate([3, 1, 4, 2, 5, 1, 6]))   # 1
print(list_first_duplicate([7, 8, 9, 10]))           # -1
print(list_first_duplicate([4, 5, 6, 4, 6]))         # 4
````

-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
# Using a set
from typing import List

def set_first_duplicate(nums: List[int]) -> int:
    seen = set()
    for n in nums:
        if n in seen:
            return n
        seen.add(n)
    return -1
````

````python
print(set_first_duplicate([3, 1, 4, 2, 5, 1, 6]))   # 1
print(set_first_duplicate([7, 8, 9, 10]))           # -1
print(set_first_duplicate([4, 5, 6, 4, 6]))         # 4
````


-   Explain why your solution works


Using [seen], check if you have seen a number before. The loop goes through the list from left to right. This make sure that the first time you detect a repeat, must be the earliest duplicate in the set. The worst case is to scan the whole set once O(1). After scanning the entire set, if no repeat number occur, return -1.


-   Explain the problem’s time and space complexity


This is the best time and space complexity solution.

Time complexity O(n) = O(1): You touch each element once, and each lookup is constant time.

Space complexity O(n) = O(1): The worst case is no repeat number were found which mean you need to store every number once.


-   Explain the thinking to an alternative solution (no coding required, but a classmate reading this should be able to code it up based off your text)


The alternative solution is to use a dictionary to keep track of which numbers you have already seen when you scan the list.

First, create an empty dictionary and loop through the list. For each number, if it is not in the dictionary, add it with a count of 1. If it is in the dictionary, add 1. at the end, it will tell you exactly how many times each number appears.

Then, find the first number that appears more than once. Loop through the original list again, from left to right. For each number, check its count in the dictionary. The first number count >= 1 is the first number to be repeated.

At the end, if no number has a count greater than 1, no duplicates, and return –1.
